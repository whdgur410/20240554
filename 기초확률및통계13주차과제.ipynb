```python
import matplotlib.pyplot as plt
import scipy.stats as stats

# 매개변수 설정
n = 10   # 시행 횟수
p = 0.5  # 성공 확률
x_values = range(n+1)

# 확률질량함수 계산
probabilities = [stats.binom.pmf(k, n, p) for k in x_values]

# 그래프 그리기
plt.figure(figsize=(8, 6))
plt.stem(x_values, probabilities, basefmt=" ")
plt.plot(x_values, probabilities, 'o', color='red')
plt.title('Binomial Distribution: n=10, p=0.5')
plt.xlabel('Number of Successes')
plt.ylabel('Probability')
plt.show()
```

```python
# 필요한 라이브러리 설치 및 불러오기
# !pip install scipy   # 주석 처리: 이미 설치되어 있다면 실행할 필요 없음
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

# 동전 던지기 횟수 설정
n_values = [10, 20, 40, 100]
p = 0.5  # 동전 앞면이 나올 확률

# 그래프 설정
plt.figure(figsize=(14, 10))

for i, n in enumerate(n_values, 1):
    x = np.arange(0, n + 1)
    pmf_values = stats.binom.pmf(x, n, p)

    # 2x2 subplot 생성
    plt.subplot(2, 2, i)
    plt.stem(x, pmf_values, linefmt='C0-', markerfmt='C0o', basefmt=" ")
    plt.title(f'{n} Coin Flips')
    plt.xlabel('Number of Heads')
    plt.ylabel('Probability')
    plt.ylim(0, max(pmf_values) + 0.05)

plt.tight_layout()
plt.show()
```

```python
# 필요한 라이브러리 불러오기
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import matplotlib
import numpy as np
from scipy.stats import poisson

# 나눔고딕 폰트 설정
font_path = "/usr/share/fonts/truetype/nanum/NanumGothic.ttf"
font_prop = fm.FontProperties(fname=font_path)
matplotlib.rcParams['font.family'] = font_prop.get_name()
plt.rcParams['axes.unicode_minus'] = False  # 마이너스 깨짐 방지

# Poisson 분포 PMF 시각화
lambda_val = 12
k_values = np.arange(0, 31)
pmf_values = poisson.pmf(k_values, lambda_val)

plt.figure(figsize=(10, 6))
markerline, stemlines, baseline = plt.stem(k_values, pmf_values, basefmt=" ")

# 그래프 스타일 설정
plt.setp(markerline, color='green', marker='o')
plt.setp(stemlines, color='gray')
plt.setp(baseline, visible=False)

# 제목과 라벨 (한글)
plt.title('Poisson 확률질량함수(PMF)\n(도서관 하루 방문자수, λ=12)', fontproperties=font_prop)
plt.xlabel('방문자수(k)', fontproperties=font_prop)
plt.ylabel('확률 P(X=k)', fontproperties=font_prop)

# 축 및 격자 설정
plt.ylim(0, max(pmf_values) + 0.02)
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

```python
# 필요한 라이브러리 불러오기
import numpy as np
import matplotlib.pyplot as plt

# 기하분포 매개변수 설정
p = 0.1  # 성공 확률

# k 값 설정 (1일부터 100일까지)
k_values = np.arange(1, 101)

# 확률질량함수(PMF) 계산
pmf_values = (1 - p) ** (k_values - 1) * p

# 그래프 설정
plt.figure(figsize=(12, 8))
markerline, stemlines, baseline = plt.stem(k_values, pmf_values, basefmt=" ")

# 그래프 스타일
plt.setp(markerline, color='purple', marker='o')
plt.setp(stemlines, color='gray')
plt.setp(baseline, visible=False)

# y축 범위 설정
plt.ylim(0, 1)

# 제목 및 축 이름
plt.title('Geometric Probability Mass Function (PMF)\nTime to First Cyber Attack')
plt.xlabel('Days until First Successful Cyber Attack')
plt.ylabel('Probability')

# 격자 추가
plt.grid(True, linestyle='--', alpha=0.5)

# 그래프 출력
plt.show()
```
